```python
# 시도 목록
sido_list = [
    "서울특별시", "부산광역시", "대구광역시", "인천광역시",
    "광주광역시", "대전광역시", "울산광역시", "세종특별자치시",
    "경기도", "강원특별자치도", "충청북도", "충청남도",
    "전북특별자치도", "전라남도", "경상북도", "경상남도",
    "제주특별자치도"
]

# 현재 소속 시도 저장
current_sido = None
sido_data = []

for region in df["행정구역(시군구)별"]:

    if region in sido_list:
        current_sido = region

    sido_data.append(current_sido)

# 시도 컬럼 추가
df["시도"] = sido_data

# 결과 확인
print(df[["시도", "행정구역(시군구)별", "연령별"]].head(40))
```

```python
# 출장소가 포함된 지역 확인
branch_offices = df[
    df["행정구역(시군구)별"].str.contains("출장소", na=False)
]

print(
    branch_offices["행정구역(시군구)별"]
    .drop_duplicates()
    .to_list()
)

# 시도 전체 행 확인
print("\n[시도 전체 행]")
print(sido_list)
```

```python
# 제외 후보 지역 목록
exclude_sido = sido_list

exclude_branch = [
    "중구영종출장소", "중구용유출장소", "서구검단출장소",
    "북부출장소", "송탄출장소", "안중출장소", "풍양출장소",
    "화성시동부출장소", "화성시동탄출장소", "동해출장소",
    "전주시효자출장소", "익산시함열출장소",
    "사천남양출장소", "장유출장소", "양산시웅상출장소"
]

# 제외 후보 표시
df["제외후보"] = (
    df["행정구역(시군구)별"].isin(exclude_sido + exclude_branch)
)

# 결과 확인
print(df["제외후보"].value_counts())
```

```python
# 일반구 후보 목록
general_gu = [
    "장안구", "권선구", "팔달구", "영통구",
    "수정구", "중원구", "분당구",
    "만안구", "동안구",
    "원미구", "소사구", "오정구",
    "상록구", "단원구",
    "덕양구", "일산동구", "일산서구",
    "처인구", "기흥구", "수지구",
    "상당구", "서원구", "흥덕구", "청원구",
    "동남구", "서북구",
    "완산구", "덕진구",
    "의창구", "성산구", "마산합포구", "마산회원구", "진해구"
]

# 시도와 지역명 확인
gu_data = df[
    df["행정구역(시군구)별"].isin(general_gu)
]

print(
    gu_data[["시도", "행정구역(시군구)별"]]
    .drop_duplicates()
    .to_string(index=False)
)
```

```python
# 일반구도 제외 후보에 추가
df["제외후보"] = (
    df["제외후보"]
    | df["행정구역(시군구)별"].isin(general_gu)
)

# 결과 확인
print(df["제외후보"].value_counts())
```

```python
# 제외 후보 지역 목록 확인
excluded_regions = (
    df.loc[df["제외후보"], ["시도", "행정구역(시군구)별"]]
    .drop_duplicates()
)

print(excluded_regions.to_string(index=False))
print("\n제외 후보 지역 수:", len(excluded_regions))
```

```python
# 수원시와 일반구의 2025년 65세 이상 인구 확인

suwon = df[
    df["행정구역(시군구)별"].isin(
        ["수원시", "장안구", "권선구", "팔달구", "영통구"]
    )
]

suwon_pivot = suwon.pivot_table(
    index="행정구역(시군구)별",
    columns="연령별",
    values="2025 년",
    aggfunc="sum"
)

# 지역별 전체 연령 합계
print(suwon_pivot.sum(axis=1))
```

```python
# 2025년 기준 상위 시와 일반구 비교
parent_groups = {
    "성남시": ["수정구", "중원구", "분당구"],
    "안양시": ["만안구", "동안구"],
    "부천시": ["원미구", "소사구", "오정구"],
    "안산시": ["상록구", "단원구"],
    "고양시": ["덕양구", "일산동구", "일산서구"],
    "용인시": ["처인구", "기흥구", "수지구"],
    "청주시": ["상당구", "서원구", "흥덕구", "청원구"],
    "천안시": ["동남구", "서북구"],
    "전주시": ["완산구", "덕진구"],
    "창원시(통합)": [
        "의창구", "성산구", "마산합포구",
        "마산회원구", "진해구"
    ]
}

for parent, children in parent_groups.items():
    parent_total = df.loc[
        (df["행정구역(시군구)별"] == parent),
        "2025 년"
    ].sum()

    child_total = df.loc[
        df["행정구역(시군구)별"].isin(children),
        "2025 년"
    ].sum()

    print(f"\n[{parent}]")
    print(f"상위 시: {parent_total:,.0f}명")
    print(f"일반구 합계: {child_total:,.0f}명")
    print(f"차이: {parent_total - child_total:,.0f}명")
```

```python
# 제외 후보를 제거한 데이터 생성
df_clean = df[df["제외후보"] == False].copy()

# 제외 후보 표시 컬럼 제거
df_clean = df_clean.drop(columns=["제외후보"])

# 결과 확인
print("원본 데이터:", df.shape)
print("정리된 데이터:", df_clean.shape)

print("\n지역 개수:")
print(df_clean[["시도", "행정구역(시군구)별"]].drop_duplicates().shape[0])
```

```python
# 연도별 인구 컬럼
year_columns = [
    "2021 년", "2022 년", "2023 년",
    "2024 년", "2025 년"
]

# 지역별 연령 인구 합산
elderly_total = (
    df_clean
    .groupby(["시도", "행정구역(시군구)별"], as_index=False)[year_columns]
    .sum(min_count=1)
)

# 결과 확인
print(elderly_total.head(10))
print("\n데이터 크기:", elderly_total.shape)
```

```python
# 지역별 연도 데이터 결측값 확인
print(elderly_total[year_columns].isnull().sum())

# 결측값이 있는 지역 확인
missing_regions = elderly_total[
    elderly_total[year_columns].isnull().any(axis=1)
]

print("\n[결측값이 있는 지역]")
print(missing_regions.to_string(index=False))
```

```python
# 군위군 데이터 확인
gunwi = elderly_total[
    elderly_total["행정구역(시군구)별"] == "군위군"
]

print(gunwi)
```

```python
# 군위군 데이터의 연도별 인구 합산
gunwi_total = gunwi[year_columns].sum(min_count=1)

# 군위군 기존 두 행 제거
elderly_total = elderly_total[
    elderly_total["행정구역(시군구)별"] != "군위군"
].copy()

# 통합된 군위군 한 행 생성
gunwi_row = pd.DataFrame([{
    "시도": "대구광역시",
    "행정구역(시군구)별": "군위군",
    **gunwi_total.to_dict()
}])

# 기존 데이터에 통합 군위군 추가
elderly_total = pd.concat(
    [elderly_total, gunwi_row],
    ignore_index=True
)

# 결과 확인
print(elderly_total[elderly_total["행정구역(시군구)별"] == "군위군"])
print("\n최종 데이터 크기:", elderly_total.shape)
```

```python
# 연도별 결측값 확인
print(elderly_total[year_columns].isnull().sum())

# 전체 결측값 개수
print("\n전체 결측값:", elderly_total[year_columns].isnull().sum().sum())
```

```python
# 최종 데이터 CSV 저장
elderly_total.to_csv(
    "elderly_clean.csv",
    index=False,
    encoding="utf-8-sig"
)

print("elderly_clean.csv 저장 완료!")
```
