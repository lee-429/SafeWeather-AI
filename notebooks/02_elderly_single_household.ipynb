```python
import pandas as pd
```

```python
# CSV 파일 불러오기
df = pd.read_csv(
    "지역별 노인 1인 가구 수.csv",
    encoding="cp949"
)
```

```python
# 데이터 기본 정보 확인
print(df.head(10))
print("\n데이터 크기:", df.shape)
print("\n컬럼명:", df.columns.tolist())
```

```python
# 첫 번째 행 제거
df = df.iloc[1:].reset_index(drop=True)

# 데이터 확인
print(df.head(10))
print("\n데이터 크기:", df.shape)

# 연령별 항목 확인
print("\n연령 종류:")
print(df["연령별"].unique())
```

```python
# 지역별 데이터 개수 확인
print(df["행정구역별(시군구)"].value_counts().head(20))

# 전체 지역 종류 확인
regions = df["행정구역별(시군구)"].drop_duplicates()

print("\n지역 개수:", len(regions))
print(regions.to_string(index=False))
```

```python
# 시도 목록
sido_list = [
    "서울특별시", "부산광역시", "대구광역시", "인천광역시",
    "광주광역시", "대전광역시", "울산광역시", "세종특별자치시",
    "경기도", "강원특별자치도", "충청북도", "충청남도",
    "전북특별자치도", "전라남도", "경상북도", "경상남도",
    "제주특별자치도"
]

current_sido = None
sido_data = []

for region in df["행정구역별(시군구)"]:
    if region in sido_list:
        current_sido = region

    sido_data.append(current_sido)

# 시도 컬럼 추가
df["시도"] = sido_data

# 결과 확인
print(df.head(20))
```

```python
# 지역별 연령 데이터 개수 확인
print(df.groupby(
    ["시도", "행정구역별(시군구)"]
)["연령별"].count().value_counts())

# 전국 행 확인
print("\n전국 데이터:")
print(df[df["행정구역별(시군구)"] == "전국"])
```

```python
# 전국 및 시도 합계 제외
df_clean = df[
    ~df["행정구역별(시군구)"].isin(
        ["전국"] + sido_list
    )
].copy()

# 결과 확인
print("기존 데이터:", df.shape)
print("전국 및 시도 합계 제외:", df_clean.shape)

print("\n시도별 지역 확인:")
print(df_clean["시도"].value_counts(dropna=False))
```

```python
# 일반구 목록
general_gu = [
    "장안구", "권선구", "팔달구", "영통구",
    "수정구", "중원구", "분당구",
    "만안구", "동안구",
    "원미구", "소사구", "오정구",
    "상록구", "단원구",
    "덕양구", "일산동구", "일산서구",
    "처인구", "기흥구", "수지구",
    "상당구", "서원구", "흥덕구", "청원구",
    "동남구", "서북구",
    "완산구", "덕진구",
    "의창구", "성산구", "마산합포구", "마산회원구", "진해구"
]

# 일반구 제외
df_clean = df_clean[
    ~df_clean["행정구역별(시군구)"].isin(general_gu)
].copy()

# 결과 확인
print("정리된 데이터:", df_clean.shape)

print("\n일반구 제외 후 지역 수:")
print(df_clean[["시도", "행정구역별(시군구)"]].drop_duplicates().shape[0])
```

```python
# 65세 이상 연령 구간
elderly_ages = [
    "65~69세",
    "70~74세",
    "75~79세",
    "80~84",
    "85세이상"
]

# 65세 이상 데이터만 추출
elderly_household = df_clean[
    df_clean["연령별"].isin(elderly_ages)
].copy()

# 연도별 숫자 컬럼
year_columns = ["2021", "2022", "2023", "2024", "2025"]

# 지역별 노인 1인 가구 수 합산
elderly_household_total = (
    elderly_household
    .groupby(["시도", "행정구역별(시군구)"], as_index=False)[year_columns]
    .sum(min_count=1)
)

# 결과 확인
print(elderly_household_total.head(10))
print("\n데이터 크기:", elderly_household_total.shape)
```

```python
# 연도별 컬럼의 데이터 타입 확인
print(df_clean[year_columns].dtypes)
```

```python
# 연도별 컬럼을 문자열에서 숫자형으로 변환
df_clean[year_columns] = df_clean[year_columns].apply(
    pd.to_numeric,
    errors="coerce"
)

# 데이터 타입 확인
print(df_clean[year_columns].dtypes)
```

```python
elderly_household = df_clean[
    df_clean["연령별"].isin(elderly_ages)
].copy()
```

```python
elderly_household_total = (
    elderly_household
    .groupby(["시도", "행정구역별(시군구)"], as_index=False)[year_columns]
    .sum(min_count=1)
)

print(elderly_household_total.head(10))
print("\n데이터 크기:", elderly_household_total.shape)
```

```python
elderly_total = pd.read_csv(
    "65세이상_고령인구_정제.csv",
    encoding="utf-8-sig"
)

print(elderly_total.shape)
```

```python
elderly_regions = set(
    zip(elderly_total["시도"], elderly_total["행정구역(시군구)별"])
)

household_regions = set(
    zip(
        elderly_household_total["시도"],
        elderly_household_total["행정구역별(시군구)"]
    )
)

print("고령인구에만 있는 지역:")
print(elderly_regions - household_regions)

print("\n노인 1인 가구에만 있는 지역:")
print(household_regions - elderly_regions)
```

```python
sejong = df_elderly_original[
    df_elderly_original["행정구역(시군구)별"] == "세종특별자치시"
].copy()

print(sejong.shape)
```

```python
year_columns = [
    "2021 년", "2022 년", "2023 년", "2024 년", "2025 년"
]

sejong_total = (
    sejong[year_columns]
    .sum()
)

print(sejong_total)
```

```python
sejong_row = pd.DataFrame([{
    "시도": "세종특별자치시",
    "행정구역(시군구)별": "세종시",
    **sejong_total.to_dict()
}])

elderly_total = pd.concat(
    [elderly_total, sejong_row],
    ignore_index=True
)

print(elderly_total.shape)
```

```python
elderly_total["행정구역(시군구)별"] = (
    elderly_total["행정구역(시군구)별"]
    .replace("창원시(통합)", "창원시")
)
```

```python
year_columns = ["2021", "2022", "2023", "2024", "2025"]
```

```python
gunwi = elderly_household_total[
    elderly_household_total["행정구역별(시군구)"] == "군위군"
]

gunwi_total = gunwi[year_columns].sum(min_count=1)

# 기존 군위군 행 제거
elderly_household_total = elderly_household_total[
    elderly_household_total["행정구역별(시군구)"] != "군위군"
].copy()

# 대구광역시 군위군으로 통합
gunwi_row = pd.DataFrame([{
    "시도": "대구광역시",
    "행정구역별(시군구)": "군위군",
    **gunwi_total.to_dict()
}])

elderly_household_total = pd.concat(
    [elderly_household_total, gunwi_row],
    ignore_index=True
)

print(elderly_household_total.shape)
```

```python
elderly_regions = set(
    zip(
        elderly_total["시도"],
        elderly_total["행정구역(시군구)별"]
    )
)

household_regions = set(
    zip(
        elderly_household_total["시도"],
        elderly_household_total["행정구역별(시군구)"]
    )
)

print("고령인구에만 있는 지역:")
print(elderly_regions - household_regions)

print("\n노인 1인 가구에만 있는 지역:")
print(household_regions - elderly_regions)
```
