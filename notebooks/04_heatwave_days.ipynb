```python
import pandas as pd
import numpy as np

df = pd.read_csv("폭염일수.csv", encoding="cp949")

print("데이터 크기:", df.shape)
display(df.head(15))

print("\n컬럼명:")
print(df.columns.tolist())

print("\n마지막 5행:")
display(df.tail(5))
```

```python
heat = df.iloc[34:, :3].copy()
heat.columns = ["연도", "날짜", "지점명"]

print("크기:", heat.shape)
display(heat.head())
display(heat.tail())
print(heat.dtypes)
```

```python
heat["일시"] = pd.to_datetime(heat["연도"] + heat["날짜"], format="%Y%m%d", errors="coerce")

print("날짜 변환 실패 개수:", heat["일시"].isna().sum())
print("기간:", heat["일시"].min(), "~", heat["일시"].max())

print("\n중복 행 개수:", heat.duplicated().sum())
display(heat[heat.duplicated(keep=False)])
```

```python
heat_clean = heat.drop_duplicates().copy()

print("정리 전:", len(heat), "행")
print("정리 후:", len(heat_clean), "행")
print("남은 중복:", heat_clean.duplicated().sum())

display(heat_clean[(heat_clean["지점명"] == "인제") & (heat_clean["연도"] == "2025")].head(3))
```

```python
heat_count = heat_clean.groupby(["지점명", "연도"]).size().reset_index(name="폭염일수")

print("크기:", heat_count.shape)
display(heat_count.head(10))

print("\n폭염일수 합계:", heat_count["폭염일수"].sum())
print("지점 개수:", heat_count["지점명"].nunique())
```

```python
temp = pd.read_csv("기상_일별_최고최저기온_정제.csv", encoding="utf-8-sig")
temp["일시"] = pd.to_datetime(temp["일시"])
temp["연도"] = temp["일시"].dt.year.astype(str)

hot = temp[temp["최고기온(°C)"] >= 33]
temp_count = hot.groupby(["지점명", "연도"]).size().reset_index(name="기온파일_폭염일수")

print("기온 파일 기준 크기:", temp_count.shape)
print("지점 개수:", temp_count["지점명"].nunique())

check = heat_count.merge(temp_count, on=["지점명", "연도"], how="inner")
print("\n비교 행 수:", len(check))
print("불일치 행 수:", (check["폭염일수"] != check["기온파일_폭염일수"]).sum())
```

```python
obs = temp.groupby(["지점", "지점명", "시도", "연도"]).size().reset_index(name="관측일수")

heat_final = obs.merge(temp_count, on=["지점명", "연도"], how="left")
heat_final["기온파일_폭염일수"] = heat_final["기온파일_폭염일수"].fillna(0).astype(int)
heat_final = heat_final.rename(columns={"기온파일_폭염일수": "폭염일수"})

print("크기:", heat_final.shape)
print("지점 개수:", heat_final["지점명"].nunique())
print("폭염일수 0인 행:", (heat_final["폭염일수"] == 0).sum())
display(heat_final[heat_final["폭염일수"] == 0].head(10))
```

```python
heat_final["연일수"] = heat_final["연도"].apply(lambda y: 366 if int(y) % 4 == 0 else 365)
heat_final["부족일수"] = heat_final["연일수"] - heat_final["관측일수"]

short = heat_final[heat_final["부족일수"] > 0]

print("관측일수가 부족한 행:", len(short))
print("그중 10일 이상 부족:", (short["부족일수"] >= 10).sum())
display(short.sort_values("부족일수", ascending=False).head(10))
```

```python
summer = temp[temp["일시"].dt.month.between(6, 9)]
summer_obs = summer.dropna(subset=["최고기온(°C)"]).groupby(["지점명", "연도"]).size().reset_index(name="여름관측일수")

heat_final = heat_final.merge(summer_obs, on=["지점명", "연도"], how="left")
heat_final["여름부족일수"] = 122 - heat_final["여름관측일수"]

check_summer = heat_final[heat_final["여름부족일수"] > 0]

print("여름철 관측이 부족한 행:", len(check_summer))
display(check_summer.sort_values("여름부족일수", ascending=False).head(10)[["지점명", "연도", "폭염일수", "여름관측일수", "여름부족일수"]])
```

```python
heat_final["여름관측_주의"] = heat_final["여름부족일수"] >= 5

print("주의 표시된 행:", heat_final["여름관측_주의"].sum())
display(heat_final[heat_final["여름관측_주의"]][["지점명", "연도", "폭염일수", "여름관측일수", "여름부족일수"]])
```

```python
final_cols = ["지점", "지점명", "시도", "연도", "폭염일수", "관측일수", "여름관측일수", "여름부족일수", "여름관측_주의"]
heat_out = heat_final[final_cols].sort_values(["지점", "연도"]).reset_index(drop=True)

print("크기:", heat_out.shape)
print("지점 개수:", heat_out["지점명"].nunique())
print("(지점, 연도) 중복:", heat_out.duplicated(["지점", "연도"]).sum())
print("결측치 개수:\n", heat_out.isna().sum())
print("폭염일수 합계:", heat_out["폭염일수"].sum())
print("폭염일수 최소/최대:", heat_out["폭염일수"].min(), heat_out["폭염일수"].max())

heat_out.to_csv("폭염일수_정제.csv", index=False, encoding="utf-8-sig")
print("\n저장 완료")
```
