```python
# 주소에서 시도와 시군구를 추출하는 함수
def extract_region(address):
    if pd.isna(address):
        return pd.Series([pd.NA, pd.NA])

    parts = str(address).strip().split()

    if len(parts) == 0:
        return pd.Series([pd.NA, pd.NA])

    sido = parts[0]

    # 세종특별자치시 예외 처리
    if sido == "세종특별자치시":
        sigungu = "세종시"
    elif len(parts) >= 2:
        sigungu = parts[1]
    else:
        sigungu = pd.NA

    return pd.Series([sido, sigungu])


# 무더위쉼터 지역 컬럼 생성
heat_df[["시도", "시군구"]] = heat_df["RN_DTL_ADRES"].apply(
    extract_region
)

# 한파쉼터 지역 컬럼 생성
cold_df[["시도", "시군구"]] = cold_df["RONA_DADDR"].apply(
    extract_region
)
```

```python
import numpy as np
import pandas as pd

# 주소가 있는 무더위쉼터
heat_valid = heat_df[
    heat_df["RN_DTL_ADRES"].notna()
].copy()

# 주소가 없는 한파쉼터
cold_missing = cold_df[
    cold_df["RONA_DADDR"].isna()
].copy()

# 좌표 숫자 변환
for col in ["LA", "LO"]:
    heat_valid[col] = pd.to_numeric(
        heat_valid[col], errors="coerce"
    )

for col in ["LAT", "LOT"]:
    cold_missing[col] = pd.to_numeric(
        cold_missing[col], errors="coerce"
    )

# 유효한 좌표만 사용
heat_valid = heat_valid.dropna(subset=["LA", "LO"])
cold_missing = cold_missing.dropna(subset=["LAT", "LOT"])

# 시설명과 좌표를 기준으로 비교할 결과
results = []

for _, cold in cold_missing.iterrows():

    # 위도·경도 차이를 이용한 대략적인 거리 계산
    lat_diff = heat_valid["LA"] - cold["LAT"]
    lon_diff = heat_valid["LO"] - cold["LOT"]

    lat_rad = np.radians(cold["LAT"])

    distances = 111320 * np.sqrt(
        lat_diff**2 +
        (lon_diff * np.cos(lat_rad))**2
    )

    nearest_idx = distances.idxmin()
    nearest = heat_valid.loc[nearest_idx]

    results.append({
        "한파쉼터명": cold["REARE_NM"],
        "무더위쉼터명": nearest["RSTR_NM"],
        "시도": nearest["시도"],
        "시군구": nearest["시군구"],
        "거리(m)": distances.loc[nearest_idx]
    })

cold_match = pd.DataFrame(results)

print("비교 대상:", len(cold_match))
print("100m 이내:", (cold_match["거리(m)"] <= 100).sum())
print("50m 이내:", (cold_match["거리(m)"] <= 50).sum())

display(
    cold_match.sort_values("거리(m)").head(20)
)
```

```python
# 시설명 공백 제거 후 비교
cold_match["한파쉼터명_정규화"] = (
    cold_match["한파쉼터명"]
    .astype(str)
    .str.replace(r"\s+", "", regex=True)
)

cold_match["무더위쉼터명_정규화"] = (
    cold_match["무더위쉼터명"]
    .astype(str)
    .str.replace(r"\s+", "", regex=True)
)

# 이름 일치 여부
cold_match["시설명_일치"] = (
    cold_match["한파쉼터명_정규화"]
    == cold_match["무더위쉼터명_정규화"]
)

# 50m 이내이면서 이름도 일치하는 시설
strong_matches = cold_match[
    (cold_match["거리(m)"] <= 50)
    & (cold_match["시설명_일치"])
].copy()

print("50m 이내:", (cold_match["거리(m)"] <= 50).sum())
print("50m 이내 + 시설명 일치:", len(strong_matches))

display(
    strong_matches[
        [
            "한파쉼터명",
            "무더위쉼터명",
            "시도",
            "시군구",
            "거리(m)"
        ]
    ].sort_values("거리(m)")
)
```

```python
# 시설명이 일치하는 사례만 대상으로 거리별 집계
name_matches = cold_match[
    cold_match["시설명_일치"]
].copy()

thresholds = [1, 5, 10, 20, 30, 50]

for distance in thresholds:
    count = (
        name_matches["거리(m)"] <= distance
    ).sum()

    print(f"{distance}m 이내: {count}개")

print("\n===== 시설명 일치 사례 전체 =====")

display(
    name_matches[
        [
            "한파쉼터명",
            "무더위쉼터명",
            "시도",
            "시군구",
            "거리(m)"
        ]
    ].sort_values("거리(m)")
)
```

```python
# 주소가 없는 한파쉼터 원본 인덱스 연결
cold_match["원본인덱스"] = cold_missing.index.to_numpy()

# 시설명 일치 + 좌표 10m 이내
candidates = cold_match[
    (cold_match["시설명_일치"])
    & (cold_match["거리(m)"] <= 10)
].copy()

print("지역 정보 보완 후보:", len(candidates))

display(
    candidates[
        [
            "원본인덱스",
            "한파쉼터명",
            "무더위쉼터명",
            "시도",
            "시군구",
            "거리(m)"
        ]
    ].sort_values("거리(m)")
)
```

```python
# 14개 후보의 지역 정보를 원본 한파쉼터 데이터에 반영
for idx, row in candidates.iterrows():
    original_idx = row["원본인덱스"]

    cold_df.loc[original_idx, "시도"] = row["시도"]
    cold_df.loc[original_idx, "시군구"] = row["시군구"]

print("지역 정보 보완 완료:", len(candidates), "개")

print("\n===== 지역 정보 결측 현황 =====")
print(cold_df[["시도", "시군구"]].isna().sum())
```

```python
# 시설 고유번호를 문자열로 변환
cold_df["시설번호_문자열"] = (
    cold_df["REARE_FCLT_NO"]
    .astype("string")
    .str.replace(r"\.0$", "", regex=True)
)

# 시설번호 앞 5자리 추출
cold_df["시설번호_앞5자리"] = (
    cold_df["시설번호_문자열"].str[:5]
)

# 주소 또는 기존 매칭으로 지역이 확인된 시설만 사용
known_cold = cold_df[
    cold_df["시도"].notna()
    & cold_df["시군구"].notna()
].copy()

# 앞 5자리별 지역 정보 집계
cold_prefix_check = (
    known_cold.groupby("시설번호_앞5자리")
    .agg(
        시도_종류=("시도", "nunique"),
        시군구_종류=("시군구", "nunique"),
        시도=("시도", "first"),
        시군구=("시군구", "first"),
        시설수=("REARE_FCLT_NO", "count")
    )
    .reset_index()
)

# 서로 다른 지역이 같은 코드에 연결되는지 확인
prefix_conflicts = cold_prefix_check[
    (cold_prefix_check["시도_종류"] > 1)
    | (cold_prefix_check["시군구_종류"] > 1)
]

print("지역 정보가 확인된 시설:", len(known_cold))
print("시설번호 앞 5자리 종류:", len(cold_prefix_check))
print("지역 코드 충돌:", len(prefix_conflicts))

display(prefix_conflicts.head(20))
```

```python
# 기존 지역 정보가 확인된 시설의 코드-지역 매핑
facility_prefix_map = cold_prefix_check.set_index(
    "시설번호_앞5자리"
)[["시도", "시군구"]]

# 지역 정보가 없는 시설
cold_remaining = cold_df[
    cold_df["시군구"].isna()
].copy()

# 시설번호 앞 5자리로 매핑
cold_remaining["매핑_시도"] = (
    cold_remaining["시설번호_앞5자리"]
    .map(facility_prefix_map["시도"])
)

cold_remaining["매핑_시군구"] = (
    cold_remaining["시설번호_앞5자리"]
    .map(facility_prefix_map["시군구"])
)

# 매핑 성공 여부
matched_mask = cold_remaining["매핑_시군구"].notna()

print("남은 시설:", len(cold_remaining))
print("지역 연결 가능:", matched_mask.sum())
print("연결 실패:", (~matched_mask).sum())

print("\n===== 연결 실패 시설 샘플 =====")

display(
    cold_remaining.loc[
        ~matched_mask,
        [
            "REARE_NM",
            "시설번호_문자열",
            "시설번호_앞5자리",
            "LAT",
            "LOT"
        ]
    ].head(20)
)
```

```python
# 기존 지역 정보가 없는 한파쉼터만 선택
missing_mask = cold_df["시군구"].isna()

# 시설번호 앞 5자리로 지역 정보 매핑
cold_df.loc[missing_mask, "시도"] = (
    cold_df.loc[missing_mask, "시설번호_앞5자리"]
    .map(facility_prefix_map["시도"])
)

cold_df.loc[missing_mask, "시군구"] = (
    cold_df.loc[missing_mask, "시설번호_앞5자리"]
    .map(facility_prefix_map["시군구"])
)

print("===== 최종 지역 정보 결측 현황 =====")
print(cold_df[["시도", "시군구"]].isna().sum())

print("\n전체 한파쉼터:", len(cold_df))
print("지역 정보 확보:", cold_df["시군구"].notna().sum())
print("지역 정보 결측:", cold_df["시군구"].isna().sum())
```

```python
cold_region_count = (
    cold_df.groupby(["시도", "시군구"])
    .size()
    .reset_index(name="한파쉼터_개수")
    .sort_values(["시도", "시군구"])
)

print("시군구 조합 수:", len(cold_region_count))
print("전체 한파쉼터:", cold_region_count["한파쉼터_개수"].sum())

display(cold_region_count.head(20))
```

```python
# 무더위쉼터와 한파쉼터 지역별 집계 데이터 병합
shelter_region = pd.merge(
    heat_region_count,
    cold_region_count,
    on=["시도", "시군구"],
    how="outer"
)

# 쉼터가 없는 지역은 0으로 처리
shelter_region[
    ["무더위쉼터_개수", "한파쉼터_개수"]
] = shelter_region[
    ["무더위쉼터_개수", "한파쉼터_개수"]
].fillna(0)

# 개수 컬럼 정수형 변환
shelter_region[
    ["무더위쉼터_개수", "한파쉼터_개수"]
] = shelter_region[
    ["무더위쉼터_개수", "한파쉼터_개수"]
].astype(int)

# 시도, 시군구 순으로 정렬
shelter_region = shelter_region.sort_values(
    ["시도", "시군구"]
).reset_index(drop=True)

print("===== 쉼터 통합 데이터 =====")
print("시군구 조합 수:", len(shelter_region))
print("무더위쉼터 총합:", shelter_region["무더위쉼터_개수"].sum())
print("한파쉼터 총합:", shelter_region["한파쉼터_개수"].sum())

display(shelter_region.head(20))
```

```python
# 시도 코드 매핑
sido_code_map = {
    "11": "서울특별시",
    "21": "부산광역시",
    "22": "대구광역시",
    "23": "인천광역시",
    "24": "광주광역시",
    "25": "대전광역시",
    "26": "울산광역시",
    "29": "세종특별자치시",
    "31": "경기도",
    "32": "강원특별자치도",
    "33": "충청북도",
    "34": "충청남도",
    "35": "전북특별자치도",
    "36": "전라남도",
    "37": "경상북도",
    "38": "경상남도",
    "39": "제주특별자치도"
}

boundary_work = boundary_df.copy()

boundary_work["시도코드"] = (
    boundary_work["SIGUNGU_CD"].astype(str).str[:2]
)

boundary_work["시도"] = (
    boundary_work["시도코드"].map(sido_code_map)
)

boundary_work["시군구"] = (
    boundary_work["SIGUNGU_NM"].str.split().str[0]
)

# 일반구를 상위 시군구로 통합
boundary_merged = boundary_work.dissolve(
    by=["시도", "시군구"],
    as_index=False
)

# 인구 데이터와 동일한 시도명으로 통일
boundary_merged["시도"] = boundary_merged["시도"].replace({
    "광주광역시": "전남광주통합특별시",
    "전라남도": "전남광주통합특별시"
})

print("통합된 경계 수:", len(boundary_merged))
print("지역 중복:", boundary_merged.duplicated(
    ["시도", "시군구"]
).sum())
print("결측치:")
print(boundary_merged[["시도", "시군구"]].isna().sum())
```

```python
# 무더위쉼터 지역별 개수 집계
heat_region_count = (
    heat_joined_final
    .groupby(["시도", "시군구"])
    .size()
    .reset_index(name="무더위쉼터_개수")
)

print("무더위쉼터 지역 수:", len(heat_region_count))
print("무더위쉼터 총합:", heat_region_count["무더위쉼터_개수"].sum())

display(heat_region_count.head(10))
```

```python
# 한파쉼터 지역별 개수 집계
cold_region_count = (
    cold_joined_final
    .groupby(["시도", "시군구"])
    .size()
    .reset_index(name="한파쉼터_개수")
)

print("한파쉼터 지역 수:", len(cold_region_count))
print("한파쉼터 총합:", cold_region_count["한파쉼터_개수"].sum())

display(cold_region_count.head(10))
```

```python
# 무더위쉼터와 한파쉼터 통합
shelter_region_final = pd.merge(
    heat_region_count,
    cold_region_count,
    on=["시도", "시군구"],
    how="outer",
    validate="one_to_one"
)

# 결측치 처리
shelter_region_final[
    ["무더위쉼터_개수", "한파쉼터_개수"]
] = shelter_region_final[
    ["무더위쉼터_개수", "한파쉼터_개수"]
].fillna(0).astype(int)

# 정렬
shelter_region_final = (
    shelter_region_final
    .sort_values(["시도", "시군구"])
    .reset_index(drop=True)
)

# 결과 확인
print("통합 데이터 크기:", shelter_region_final.shape)
print("지역 수:", shelter_region_final[["시도", "시군구"]].drop_duplicates().shape[0])
print("무더위쉼터 총합:", shelter_region_final["무더위쉼터_개수"].sum())
print("한파쉼터 총합:", shelter_region_final["한파쉼터_개수"].sum())

print("\n결측치:")
print(shelter_region_final.isna().sum())

display(shelter_region_final.head(10))
```

```python
# 저장 경로
PROCESSED_DIR = ROOT_DIR / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

output_path = PROCESSED_DIR / "쉼터_지역별_통합_정제.csv"

# CSV 저장
shelter_region_final.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

print("저장 완료:", output_path)
```

```python
# 무더위쉼터와 한파쉼터 개수 차이 확인
shelter_check = shelter_region_final.copy()

shelter_check["개수_차이"] = (
    shelter_check["무더위쉼터_개수"]
    - shelter_check["한파쉼터_개수"]
).abs()

# 차이가 큰 지역 상위 15개
shelter_check = shelter_check.sort_values(
    "개수_차이",
    ascending=False
)

display(shelter_check.head(15))
```
