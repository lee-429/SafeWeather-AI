```python
import pandas as pd
from pathlib import Path

# 두 CSV 불러오기
ROOT_DIR = Path.cwd().parent
PROCESSED_DIR = ROOT_DIR / "data" / "processed"

elderly_df = pd.read_csv(
    PROCESSED_DIR / "65세이상_고령인구_정제.csv"
)

single_df = pd.read_csv(
    PROCESSED_DIR / "노인_1인가구_정제.csv"
)

shelter_df = pd.read_csv(
    PROCESSED_DIR / "쉼터_지역별_통합_정제.csv"
)
```

```python
# 컬럼명 통일하기
elderly_df = elderly_df.rename(columns={
    "행정구역(시군구)별": "시군구"
})

single_df = single_df.rename(columns={
    "행정구역별(시군구)": "시군구"
})
```

```python
# 행정구역 명칭 통일하기
def normalize_region(df):
    df = df.copy()

    # 세종시 명칭 통일
    df["시군구"] = df["시군구"].replace({
        "세종특별자치시": "세종시"
    })

    # 창원시 명칭 통일
    df["시군구"] = df["시군구"].replace({
        "창원시(통합)": "창원시"
    })

    return df


elderly_df = normalize_region(elderly_df)
single_df = normalize_region(single_df)
```

```python
# 데이터별 지역 개수 확인하기
print("고령인구 지역 개수:", len(elderly_df))
print("노인 1인가구 지역 개수:", len(single_df))
print("쉼터 지역 개수:", len(shelter_df))
```

```python
# 서로 일치하지 않는 지역 찾기
elderly_regions = set(
    zip(elderly_df["시도"], elderly_df["시군구"])
)

single_regions = set(
    zip(single_df["시도"], single_df["시군구"])
)

shelter_regions = set(
    zip(shelter_df["시도"], shelter_df["시군구"])
)

print("고령인구에만 있는 지역:")
print(sorted(elderly_regions - single_regions))

print("\n노인 1인가구에만 있는 지역:")
print(sorted(single_regions - elderly_regions))

print("\n고령인구에만 있고 쉼터에는 없는 지역:")
print(sorted(elderly_regions - shelter_regions))

print("\n쉼터에만 있고 고령인구에는 없는 지역:")
print(sorted(shelter_regions - elderly_regions))
```

```python
# 중복 지역 확인하기
print("고령인구 중복 지역:")
print(elderly_df[
    elderly_df.duplicated(["시도", "시군구"], keep=False)
])

print("\n노인 1인가구 중복 지역:")
print(single_df[
    single_df.duplicated(["시도", "시군구"], keep=False)
])

print("\n쉼터 중복 지역:")
print(shelter_df[
    shelter_df.duplicated(["시도", "시군구"], keep=False)
])
```

```python
def normalize_sido(df):
    df = df.copy()

    df["시도"] = df["시도"].replace({
        "광주광역시": "전남광주통합특별시",
        "전라남도": "전남광주통합특별시"
    })

    return df


elderly_df = normalize_sido(elderly_df)
single_df = normalize_sido(single_df)
```

```python
elderly_regions = set(
    zip(elderly_df["시도"], elderly_df["시군구"])
)

shelter_regions = set(
    zip(shelter_df["시도"], shelter_df["시군구"])
)

print("고령인구에만 있는 지역:")
print(sorted(elderly_regions - shelter_regions))

print("\n쉼터에만 있는 지역:")
print(sorted(shelter_regions - elderly_regions))
```

```python
# 고령인구와 쉼터의 공통 행정구역 추출
common_regions = elderly_regions & shelter_regions

# 쉼터 데이터에서 공통 행정구역만 유지
shelter_df = shelter_df[
    shelter_df.apply(
        lambda row: (row["시도"], row["시군구"]) in common_regions,
        axis=1
    )
].copy()

print("기존 쉼터 지역 수:", len(shelter_regions))
print("공통 행정구역 수:", len(common_regions))
print("정리 후 쉼터 지역 수:", len(shelter_df))
```

```python
# 제외할 지역
exclude_regions = [
    ("인천광역시", "검단구"),
    ("인천광역시", "서해구"),
    ("인천광역시", "영종구"),
    ("인천광역시", "제물포구")
]

# 쉼터 데이터에서 4개 지역 제거
exclude_mask = pd.MultiIndex.from_frame(
    shelter_df[["시도", "시군구"]]
).isin(exclude_regions)

shelter_df = shelter_df.loc[~exclude_mask].copy()

# 결과 확인
print("제거 후 쉼터 지역 수:", len(shelter_df))
```

```python
output_path = (
    PROCESSED_DIR / "쉼터_지역별_통합_정제.csv"
)

shelter_df.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

print("저장 완료:", output_path)
```

```python
# 데이터 통합 전 최종 검증
import pandas as pd
from pathlib import Path

ROOT_DIR = Path.cwd().parent
PROCESSED_DIR = ROOT_DIR / "data" / "processed"

# 6개 데이터 불러오기
elderly_df = pd.read_csv(
    PROCESSED_DIR / "65세이상_고령인구_정제.csv"
)

single_df = pd.read_csv(
    PROCESSED_DIR / "노인_1인가구_정제.csv"
)

temperature_df = pd.read_csv(
    PROCESSED_DIR / "기상_일별_최고최저기온_정제.csv"
)

heatwave_df = pd.read_csv(
    PROCESSED_DIR / "폭염일수_정제.csv"
)

coldwave_df = pd.read_csv(
    PROCESSED_DIR / "한파일수_정제.csv"
)

shelter_df = pd.read_csv(
    PROCESSED_DIR / "쉼터_지역별_통합_정제.csv"
)

print("6개 데이터 로드 완료!")
```

```python
datasets = {
    "고령인구": elderly_df,
    "노인 1인가구": single_df,
    "일별 최고최저기온": temperature_df,
    "폭염일수": heatwave_df,
    "한파일수": coldwave_df,
    "쉼터": shelter_df
}

for name, df in datasets.items():
    print(f"\n{'=' * 40}")
    print(f"[{name}]")
    print("데이터 크기:", df.shape)
    print("컬럼:", df.columns.tolist())
    print("전체 결측치:", df.isna().sum().sum())
    print("중복 행:", df.duplicated().sum())
```

```python
# 지역별 데이터 컬럼명 통일
elderly_df = elderly_df.rename(columns={
    "행정구역(시군구)별": "시군구"
})

single_df = single_df.rename(columns={
    "행정구역별(시군구)": "시군구"
})

# 지역 조합 생성
elderly_regions = set(
    zip(elderly_df["시도"], elderly_df["시군구"])
)

single_regions = set(
    zip(single_df["시도"], single_df["시군구"])
)

shelter_regions = set(
    zip(shelter_df["시도"], shelter_df["시군구"])
)

print("고령인구 지역 수:", len(elderly_regions))
print("노인 1인가구 지역 수:", len(single_regions))
print("쉼터 지역 수:", len(shelter_regions))

print("\n고령인구와 노인 1인가구 일치:",
      elderly_regions == single_regions)

print("고령인구와 쉼터 일치:",
      elderly_regions == shelter_regions)
```

```python
def normalize_population(df):
    df = df.copy()

    # 컬럼명 통일
    df = df.rename(columns={
        "행정구역(시군구)별": "시군구",
        "행정구역별(시군구)": "시군구"
    })

    # 시도명 통일
    df["시도"] = df["시도"].replace({
        "광주광역시": "전남광주통합특별시",
        "전라남도": "전남광주통합특별시"
    })

    # 시군구명 통일
    df["시군구"] = df["시군구"].replace({
        "세종특별자치시": "세종시",
        "창원시(통합)": "창원시"
    })

    return df


elderly_df = normalize_population(elderly_df)
single_df = normalize_population(single_df)
```

```python
elderly_regions = set(
    zip(elderly_df["시도"], elderly_df["시군구"])
)

single_regions = set(
    zip(single_df["시도"], single_df["시군구"])
)

shelter_regions = set(
    zip(shelter_df["시도"], shelter_df["시군구"])
)

print("고령인구에만 있는 지역:")
print(sorted(elderly_regions - single_regions))

print("\n노인 1인가구에만 있는 지역:")
print(sorted(single_regions - elderly_regions))

print("\n고령인구에만 있는 지역 (쉼터 비교):")
print(sorted(elderly_regions - shelter_regions))

print("\n쉼터에만 있는 지역:")
print(sorted(shelter_regions - elderly_regions))

print("\n지역 개수")
print("고령인구:", len(elderly_regions))
print("노인 1인가구:", len(single_regions))
print("쉼터:", len(shelter_regions))
```

```python
print(temperature_df.isna().sum())

print("\n결측치가 있는 행:")
display(
    temperature_df[
        temperature_df.isna().any(axis=1)
    ]
)
```

```python
# 관측소별 연도별 데이터 건수
temperature_df["연도"] = pd.to_datetime(
    temperature_df["일시"]
).dt.year

station_year_count = (
    temperature_df
    .groupby(["지점", "지점명", "시도", "연도"])
    .size()
    .reset_index(name="데이터_건수")
)

print("관측소 수:", temperature_df["지점"].nunique())
print("관측소-연도 조합 수:", len(station_year_count))

print("\n연도별 관측 데이터 건수")
display(
    station_year_count
    .groupby("연도")["데이터_건수"]
    .agg(["count", "min", "max", "mean"])
)
```

```python
heat_stations = set(heatwave_df["지점"])
cold_stations = set(coldwave_df["지점"])
temp_stations = set(temperature_df["지점"])

print("기온 데이터 관측소:", len(temp_stations))
print("폭염 데이터 관측소:", len(heat_stations))
print("한파 데이터 관측소:", len(cold_stations))

print("\n기온 데이터에만 있는 관측소:")
print(sorted(temp_stations - heat_stations))

print("\n폭염 데이터에만 있는 관측소:")
print(sorted(heat_stations - temp_stations))

print("\n한파 데이터에만 있는 관측소:")
print(sorted(cold_stations - temp_stations))
```

```python
# 연도별 기대 관측일수
expected_days = {
    2021: 365,
    2022: 365,
    2023: 365,
    2024: 366,
    2025: 365
}

station_year_count["기대_관측일수"] = (
    station_year_count["연도"].map(expected_days)
)

station_year_count["부족일수"] = (
    station_year_count["기대_관측일수"]
    - station_year_count["데이터_건수"]
)

# 관측일수가 350일 미만인 관측소 확인
display(
    station_year_count[
        station_year_count["데이터_건수"] < 350
    ].sort_values(["연도", "데이터_건수"])
)
```

```python
# 서청주 2023년 데이터 추출
seocheongju_2023 = temperature_df[
    (temperature_df["지점"] == 181) &
    (temperature_df["연도"] == 2023)
].copy()

# 날짜별 데이터 확인
seocheongju_2023["월"] = (
    pd.to_datetime(seocheongju_2023["일시"]).dt.month
)

print("최초 관측일:", seocheongju_2023["일시"].min())
print("최종 관측일:", seocheongju_2023["일시"].max())
print("관측일수:", len(seocheongju_2023))

print("\n월별 관측일수")
display(
    seocheongju_2023
    .groupby("월")
    .size()
    .reindex(range(1, 13), fill_value=0)
    .rename("관측일수")
)
```
